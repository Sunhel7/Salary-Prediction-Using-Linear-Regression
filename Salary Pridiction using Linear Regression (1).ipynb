```python
data=pd.read_csv("Salary_data.csv")
```

```python
data.head()
```

```python
x=data["YearsExperience"]
y=data["Salary"]
```

```python
#print(x)
#print(y)
```

```python
def cost_compute(x,y,w,b):
    m=len(x)
    cost_sum=0
    for i in range(m):
        y_pred=w*x[i]+b
        cost_sum+=(y_pred-y[i])**2
        
    return cost_sum/(2*m)
```

```python
w=0.0
b=0.0
alpha=0.00001
itern=1000000
m=len(x)
j_hist=[]
```

```python
for j in range(itern):
    dw=0.0
    db=0.0
    for i in range(m):
        y_pred=w*x[i]+b
        err=y_pred-y[i]
        
        dw+=err*x[i]
        db+=err
        
    dw=dw/m
    db=db/m
    
    w=w-alpha*dw
    b=b-alpha*db
    
    if j%1000==0:
        cost=cost_compute(x,y,w,b)
        j_hist.append(cost)
        
```

```python
y_pred=w*x+b
```

```python
print("Trained w: ",w)
print("Trained b: ",b)
print("Final Cost: ",cost_compute(x,y,w,b))
```

```python

import plotly.graph_objects as go

# Predictions using trained w and b
y_pred = w * x + b

fig = go.Figure()

# Actual data points
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode="markers",
    name="Actual Salary",
    marker=dict(color="red", size=8, symbol="x")))

# Prediction line
fig.add_trace(go.Scatter(
    x=x,
    y=y_pred,
    mode="lines",
    name="Prediction Line",
    line=dict(color="blue", width=3)))

fig.update_layout(
    title="Salary vs Experience",
    xaxis_title="Years of Experience",
    yaxis_title="Salary",
    template="plotly_white",
    width=900,
    height=600)

fig.show()
```

```python

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

iterations = np.arange(len(j_hist))
start = max(0, len(j_hist) - 500)

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Cost vs Iterations (Start)",
                    "Cost vs Iterations (End)")
)

# Starting iterations
fig.add_trace(
    go.Scatter(x=iterations[:500], y=j_hist[:500],
               mode="lines", name="Start"),
    row=1, col=1
)

# Last 500 recorded costs
fig.add_trace(
    go.Scatter(x=iterations[start:], y=j_hist[start:],
               mode="lines", name="End"),
    row=1, col=2
)

fig.update_xaxes(title_text="Recorded iteration")
fig.update_yaxes(title_text="Cost")

fig.update_layout(
    title="Cost Function During Gradient Descent",
    width=1000, height=500
)

fig.show()
```
